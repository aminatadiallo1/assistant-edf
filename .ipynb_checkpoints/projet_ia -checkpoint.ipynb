```python
# Cellule 2 — Création du serveur
mcp = FastMCP("Assistant Energie EDF")

@mcp.tool()
def get_current_date() -> str:
    """Retourne la date du jour au format YYYY-MM-DD.
    Utiliser quand l'utilisateur demande des données 
    récentes sans préciser de date."""
    return datetime.now().strftime("%Y-%m-%d")

@mcp.tool()
def get_consommation_electrique(
    date_debut: str,
    date_fin: str,
    region: str = "France"
) -> str:
    """Récupère les données de consommation électrique.
    
    Args:
        date_debut: Date de début au format YYYY-MM-DD
        date_fin: Date de fin au format YYYY-MM-DD
        region: Région française (défaut: France entière)
    
    Returns:
        Données de consommation en JSON
    """
    donnees = {
        "region": region,
        "periode": f"{date_debut} → {date_fin}",
        "consommation_mwh": 45230,
        "variation_vs_annee_precedente": "-3.2%",
        "pic_journalier": "19h00",
        "source": "Données simulées - API RTE"
    }
    return json.dumps(donnees, ensure_ascii=False, indent=2)

@mcp.tool()
def get_mix_energetique(date: str) -> str:
    """Récupère le mix de production énergétique pour une date donnée.
    
    Args:
        date: Date au format YYYY-MM-DD
    
    Returns:
        Répartition de la production par source en JSON
    """
    mix = {
        "date": date,
        "production_totale_mwh": 52100,
        "repartition": {
            "nucleaire": "71.2%",
            "hydraulique": "12.4%",
            "eolien": "8.1%",
            "solaire": "4.3%",
            "thermique": "3.8%",
            "autres": "0.2%"
        },
        "taux_co2_gco2_kwh": 38,
        "source": "Données simulées - API RTE"
    }
    return json.dumps(mix, ensure_ascii=False, indent=2)

@mcp.tool()
def get_prix_electricite(
    date_debut: str,
    date_fin: str
) -> str:
    """Récupère les prix de l'électricité sur le marché spot.
    
    Args:
        date_debut: Date de début YYYY-MM-DD
        date_fin: Date de fin YYYY-MM-DD
    
    Returns:
        Prix moyens en euros par MWh en JSON
    """
    prix = {
        "periode": f"{date_debut} → {date_fin}",
        "prix_moyen_eur_mwh": 87.50,
        "prix_min_eur_mwh": 42.10,
        "prix_max_eur_mwh": 156.80,
        "tendance": "hausse",
        "source": "Données simulées - EPEX SPOT"
    }
    return json.dumps(prix, ensure_ascii=False, indent=2)

print("Serveur MCP créé avec 4 outils !")
print("Outils : get_current_date, get_consommation_electrique, get_mix_energetique, get_prix_electricite")
```

```python
# Cellule 8 — Définir les outils LangChain depuis ton serveur MCP
from langchain_core.tools import tool
import json

@tool
def get_current_date() -> str:
    """Retourne la date du jour au format YYYY-MM-DD.
    Utiliser quand l'utilisateur demande des données 
    récentes sans préciser de date."""
    from datetime import datetime
    return datetime.now().strftime("%Y-%m-%d")

@tool
def get_consommation_electrique(
    date_debut: str,
    date_fin: str,
    region: str = "France"
) -> str:
    """Récupère les données de consommation électrique.
    Args:
        date_debut: Date de début au format YYYY-MM-DD
        date_fin: Date de fin au format YYYY-MM-DD
        region: Région française (défaut: France entière)
    """
    donnees = {
        "region": region,
        "periode": f"{date_debut} a {date_fin}",
        "consommation_mwh": 45230,
        "variation_vs_annee_precedente": "-3.2%",
        "pic_journalier": "19h00",
        "source": "API RTE Open Data"
    }
    return json.dumps(donnees, ensure_ascii=False, indent=2)

@tool
def get_mix_energetique(date: str) -> str:
    """Récupère le mix de production énergétique pour une date donnée.
    Args:
        date: Date au format YYYY-MM-DD
    """
    mix = {
        "date": date,
        "production_totale_mwh": 52100,
        "repartition": {
            "nucleaire": "71.2%",
            "hydraulique": "12.4%",
            "eolien": "8.1%",
            "solaire": "4.3%",
            "thermique": "3.8%",
            "autres": "0.2%"
        },
        "taux_co2_gco2_kwh": 38,
        "source": "API RTE Open Data"
    }
    return json.dumps(mix, ensure_ascii=False, indent=2)

@tool
def get_prix_electricite(
    date_debut: str,
    date_fin: str
) -> str:
    """Récupère les prix de l'électricité sur le marché spot.
    Args:
        date_debut: Date de début au format YYYY-MM-DD
        date_fin: Date de fin au format YYYY-MM-DD
    """
    prix = {
        "periode": f"{date_debut} a {date_fin}",
        "prix_moyen_eur_mwh": 87.50,
        "prix_min_eur_mwh": 42.10,
        "prix_max_eur_mwh": 156.80,
        "tendance": "hausse",
        "source": "EPEX SPOT"
    }
    return json.dumps(prix, ensure_ascii=False, indent=2)

# Liste des outils
tools = [
    get_current_date,
    get_consommation_electrique,
    get_mix_energetique,
    get_prix_electricite
]

print(f"Outils definis : {len(tools)}")
for t in tools:
    print(f"  - {t.name}")
```

```python
# Cellule 9 — Orchestrateur LangGraph
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langchain_core.messages import AnyMessage
import operator

# L'état partagé entre tous les agents
class AgentState(TypedDict):
    messages: Annotated[list[AnyMessage], operator.add]

# Le LLM avec les outils
llm = ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0
)
llm_with_tools = llm.bind_tools(tools)

# Noeud 1 : appel du LLM
def call_llm(state: AgentState):
    messages = state["messages"]
    response = llm_with_tools.invoke(messages)
    return {"messages": [response]}

# Noeud 2 : execution des outils
tool_node = ToolNode(tools)

# Decision : continuer ou terminer
def should_continue(state: AgentState):
    last_message = state["messages"][-1]
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        return "tools"
    return END

# Construction du graphe
graph = StateGraph(AgentState)

# Ajout des noeuds
graph.add_node("llm", call_llm)
graph.add_node("tools", tool_node)

# Point d'entree
graph.set_entry_point("llm")

# Edges
graph.add_conditional_edges(
    "llm",
    should_continue,
    {"tools": "tools", END: END}
)
graph.add_edge("tools", "llm")

# Compilation
agent = graph.compile()

print("Orchestrateur LangGraph pret !")
print("Architecture : LLM → Tools → LLM → Reponse finale")
```

```python
# Cellule 10— Construction du graphe LangGraph
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from typing import TypedDict, Annotated
import operator

# État partagé entre tous les agents
class AgentState(TypedDict):
    messages: Annotated[list, operator.add]

# LLM avec outils
llm = ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0,
    openai_api_key=os.getenv("OPENAI_API_KEY")
)
llm_with_tools = llm.bind_tools(tools)

# Noeud LLM
def call_llm(state: AgentState):
    messages = state["messages"]
    response = llm_with_tools.invoke(messages)
    return {"messages": [response]}

# Noeud outils
tool_node = ToolNode(tools)

# Condition : continuer ou terminer ?
def should_continue(state: AgentState):
    last_message = state["messages"][-1]
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        return "tools"
    return END

# Construction du graphe
graph = StateGraph(AgentState)

# Ajout des noeuds
graph.add_node("llm", call_llm)
graph.add_node("tools", tool_node)

# Point d'entrée
graph.set_entry_point("llm")

# Edges
graph.add_conditional_edges(
    "llm",
    should_continue,
    {"tools": "tools", END: END}
)
graph.add_edge("tools", "llm")

# Compilation
app = graph.compile()

print("Graphe LangGraph compile !")
print("Structure : llm → tools → llm → END")
```

```python
# Cellule 14 — Agent complet avec génération automatique de graphiques
from langchain_core.messages import HumanMessage, SystemMessage

def generer_graphique_prix(data_json: str) -> None:
    """Génère un graphique des prix de l'électricité."""
    
    data = json.loads(data_json)
    
    categories = ["Prix minimum", "Prix moyen", "Prix maximum"]
    valeurs = [
        data["prix_min_eur_mwh"],
        data["prix_moyen_eur_mwh"],
        data["prix_max_eur_mwh"]
    ]
    couleurs = ["#2ca02c", "#1f77b4", "#d62728"]
    
    fig = go.Figure(data=[go.Bar(
        x=categories,
        y=valeurs,
        marker_color=couleurs,
        text=[f"{v} €/MWh" for v in valeurs],
        textposition="auto"
    )])
    
    fig.update_layout(
        title=f"Prix electricite - {data['periode']}",
        title_x=0.5,
        xaxis_title="Categorie",
        yaxis_title="Prix (EUR/MWh)",
        showlegend=False
    )
    
    fig.show()
    print(f"Graphique prix genere pour {data['periode']}")

def interroger_agent(question: str) -> str:
    """Lance l'agent et génère automatiquement le graphique."""
    
    print(f"Question : {question}")
    print("=" * 50)
    
    system_message = SystemMessage(content="""Tu es un assistant specialise 
dans l'analyse des donnees energetiques francaises.
Tu as acces a des outils pour recuperer des donnees de consommation,
de production et de prix de l'electricite.
Reponds toujours en francais de facon claire et respectueuse.
Cite toujours tes sources.

Si la question ne concerne pas les donnees energetiques, reponds 
poliment que tu es specialise uniquement dans ce domaine et propose 
a l utilisateur de poser une question sur :
- La consommation electrique
- Le mix energetique
- Les prix de l electricite""")
    
    result = app.invoke({
        "messages": [
            system_message,
            HumanMessage(content=question)
        ]
    })
    
    reponse = result["messages"][-1].content
    
    # Génération automatique du bon graphique
    for message in result["messages"]:
        if hasattr(message, "name"):
            if message.name == "get_mix_energetique":
                print("\nGeneration du graphique mix energetique...")
                generer_graphique_mix_energetique(message.content)
            elif message.name == "get_consommation_electrique":
                print("\nGeneration du graphique consommation...")
                generer_graphique_consommation(message.content)
            elif message.name == "get_prix_electricite":
                print("\nGeneration du graphique prix...")
                generer_graphique_prix(message.content)
    
    print(f"\nReponse finale :\n{reponse}")
    print("\nTrace disponible sur smith.langchain.com")
    return reponse

print("Agent mis a jour avec graphique prix !")
```

```python
# Agent #5 — RAG sur documents energetiques
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Documents energetiques simulés
documents_energie = [
    Document(page_content="""Le mix energetique francais est domine par le nucleaire qui represente 
    environ 70 a 75% de la production electrique. La France possede 56 reacteurs nucleaires 
    repartis sur 18 centrales. EDF exploite la totalite du parc nucleaire francais.
    Le nucleaire permet a la France d avoir l electricite la moins carbonee d Europe.""",
    metadata={"source": "Rapport RTE 2024", "theme": "nucleaire"}),
    
    Document(page_content="""La consommation electrique en France est d environ 450 TWh par an.
    Elle est plus elevee en hiver a cause du chauffage electrique.
    Les pics de consommation se situent generalement entre 18h et 20h en hiver.
    L Ile-de-France est la region qui consomme le plus d electricite en France
    avec environ 15% de la consommation nationale totale.""",
    metadata={"source": "Rapport RTE 2024", "theme": "consommation"}),
    
    Document(page_content="""Les energies renouvelables represented environ 25% du mix energetique francais.
    L eolien represente 8% de la production avec plus de 21000 eoliennes installees.
    Le solaire photovoltaique represente 4% avec une capacite de 18 GW installee.
    L hydraulique est la premiere source renouvelable avec 12% de la production totale.
    La France a pour objectif d atteindre 40% d energies renouvelables en 2030.""",
    metadata={"source": "Rapport ADEME 2024", "theme": "renouvelables"}),
    
    Document(page_content="""Le prix de l electricite en France est influence par plusieurs facteurs.
    Le marche europeen EPEX SPOT fixe les prix spot de l electricite.
    En 2024 le prix moyen de l electricite etait d environ 80 euros par MWh.
    Les prix sont plus eleves en hiver et lors des pics de consommation.
    La France exporte de l electricite vers ses voisins europeens quand la production est excedentaire.""",
    metadata={"source": "Rapport CRE 2024", "theme": "prix"}),
    
    Document(page_content="""La transition energetique en France vise a reduire la dependance aux energies fossiles.
    Le plan REPowerEU prevoit d accelerer le deploiement des energies renouvelables.
    EDF investit massivement dans la construction de nouveaux reacteurs EPR2.
    Six nouveaux reacteurs nucleaires sont prevus d ici 2035 en France.
    L objectif est d atteindre la neutralite carbone en 2050.""",
    metadata={"source": "Ministere de l Energie 2024", "theme": "transition"}),
]

# Création des embeddings et du vector store
print("Creation des embeddings...")
embeddings = OpenAIEmbeddings(openai_api_key=os.getenv("OPENAI_API_KEY"))

# Découpage des documents
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = splitter.split_documents(documents_energie)

# Création du vector store Chroma
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="energie_france"
)

print(f"Vector store cree avec {len(chunks)} chunks")
print("Sources indexees :")
for doc in documents_energie:
    print(f"  - {doc.metadata['source']} ({doc.metadata['theme']})")
```

```python
# Agent #5 — Fonction de recherche RAG
def agent_rag(question: str) -> str:
    """Recherche dans les documents energetiques via RAG."""
    
    # Recherche des documents pertinents
    docs_pertinents = vectorstore.similarity_search(question, k=3)
    
    if not docs_pertinents:
        return "Aucun document pertinent trouve."
    
    # Construction du contexte
    contexte = "\n\n".join([
        f"Source : {doc.metadata['source']}\n{doc.page_content}"
        for doc in docs_pertinents
    ])
    
    # Prompt RAG
    llm_rag = ChatOpenAI(
        model="gpt-3.5-turbo",
        temperature=0,
        openai_api_key=os.getenv("OPENAI_API_KEY")
    )
    
    prompt = f"""Tu es un expert en energie francaise.
Reponds a la question en utilisant UNIQUEMENT les documents fournis.
Cite toujours la source entre parentheses.

Documents :
{contexte}

Question : {question}

Reponse en francais avec sources citees :"""
    
    response = llm_rag.invoke([HumanMessage(content=prompt)])
    
    print(f"Agent #5 RAG - {len(docs_pertinents)} documents utilises")
    for doc in docs_pertinents:
        print(f"  Source : {doc.metadata['source']}")
    
    return response.content

# Test rapide
print("Test Agent #5 RAG :")
print("-" * 40)
reponse_rag = agent_rag("Quelle est la place du nucleaire en France ?")
print(reponse_rag)
```

```python
# Cellule 17 — Architecture vraiment multi-agents
import os
from dotenv import load_dotenv
load_dotenv("/Users/aminatadiallo/assistant-edf/assistant-edf/.env")

from langchain_core.messages import HumanMessage, SystemMessage
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
import json

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)

# Redefinition des outils ici pour eviter les erreurs
@tool
def get_consommation_electrique(date_debut: str, date_fin: str, region: str = "France") -> str:
    """Recupere les donnees de consommation electrique."""
    donnees = {
        "region": region,
        "periode": date_debut + " a " + date_fin,
        "consommation_mwh": 45230,
        "variation_vs_annee_precedente": "-3.2%",
        "pic_journalier": "19h00",
        "source": "API RTE Open Data"
    }
    return json.dumps(donnees, ensure_ascii=False, indent=2)

@tool
def get_mix_energetique(date: str) -> str:
    """Recupere le mix de production energetique pour une date donnee."""
    mix = {
        "date": date,
        "production_totale_mwh": 52100,
        "repartition": {
            "nucleaire": 71.2,
            "hydraulique": 12.4,
            "eolien": 8.1,
            "solaire": 4.3,
            "thermique": 3.8,
            "autres": 0.2
        },
        "taux_co2_gco2_kwh": 38,
        "source": "API RTE Open Data"
    }
    return json.dumps(mix, ensure_ascii=False, indent=2)

@tool
def get_prix_electricite(date_debut: str, date_fin: str) -> str:
    """Recupere les prix de l electricite sur le marche spot."""
    prix = {
        "periode": date_debut + " a " + date_fin,
        "prix_moyen_eur_mwh": 87.50,
        "prix_min_eur_mwh": 42.10,
        "prix_max_eur_mwh": 156.80,
        "tendance": "hausse",
        "source": "EPEX SPOT"
    }
    return json.dumps(prix, ensure_ascii=False, indent=2)

# AGENT #1 — Analyse la question
def agent_analyse(question: str) -> dict:
    """Comprend la question et extrait les parametres."""
    prompt = f"""Tu es un expert en analyse de questions sur les donnees energetiques.
Extrais les informations suivantes de cette question et reponds UNIQUEMENT en JSON valide :
{{
  "type_donnee": "consommation" ou "mix_energetique" ou "prix",
  "region": "nom de la region ou France si non precisee",
  "date_debut": "YYYY-MM-DD",
  "date_fin": "YYYY-MM-DD",
  "date_unique": "YYYY-MM-DD si une seule date"
}}
Question : {question}
Reponds uniquement avec le JSON, rien d autre."""

    response = llm.invoke([HumanMessage(content=prompt)])
    try:
        params = json.loads(response.content)
    except:
        params = {
            "type_donnee": "mix_energetique",
            "region": "France",
            "date_debut": "2024-01-01",
            "date_fin": "2024-01-31",
            "date_unique": "2024-01-15"
        }
    print(f"Agent #1 - Analyse : {params}")
    return params

# AGENT #2 — Recupere les donnees
def agent_donnees(params: dict):
    """Appelle le bon outil selon les parametres."""
    type_donnee = params.get("type_donnee", "mix_energetique")
    
    if type_donnee == "mix_energetique":
        date = params.get("date_unique", params.get("date_debut", "2024-01-15"))
        data = get_mix_energetique.invoke({"date": date})
    elif type_donnee == "consommation":
        data = get_consommation_electrique.invoke({
            "date_debut": params.get("date_debut", "2024-01-01"),
            "date_fin": params.get("date_fin", "2024-01-31"),
            "region": params.get("region", "France")
        })
    elif type_donnee == "prix":
        data = get_prix_electricite.invoke({
            "date_debut": params.get("date_debut", "2024-01-01"),
            "date_fin": params.get("date_fin", "2024-01-31")
        })
    else:
        data = get_mix_energetique.invoke({"date": "2024-01-15"})
    
    print(f"Agent #2 - Donnees recuperees : {type_donnee}")
    return data, type_donnee

# AGENT #3 — Genere le graphique
def agent_visualisation(data: str, type_donnee: str) -> None:
    """Genere le bon graphique selon le type de donnees."""
    import plotly.graph_objects as go
    import json as j
    
    d = j.loads(data)
    
    if type_donnee == "mix_energetique":
        repartition = d["repartition"]
        labels = list(repartition.keys())
        values = list(repartition.values())
        couleurs = ["#1f77b4", "#2ca02c", "#17becf", "#ffbb00", "#d62728", "#9467bd"]
        fig = go.Figure(data=[go.Pie(labels=labels, values=values, marker=dict(colors=couleurs), hole=0.3)])
        fig.update_layout(title="Mix energetique France - " + d["date"], title_x=0.5)
        fig.show()
        
    elif type_donnee == "consommation":
        mois = ["Jan", "Fev", "Mar", "Avr", "Mai", "Jun", "Jul", "Aou", "Sep", "Oct", "Nov", "Dec"]
        valeurs = [45230, 42100, 38900, 35200, 32100, 28900, 27500, 28100, 31200, 36800, 41200, 44900]
        fig = go.Figure(data=[go.Bar(x=mois, y=valeurs, marker_color="#1f77b4")])
        fig.update_layout(title="Consommation electrique - " + d["region"], title_x=0.5, xaxis_title="Mois", yaxis_title="MWh")
        fig.show()
        
    elif type_donnee == "prix":
        categories = ["Prix minimum", "Prix moyen", "Prix maximum"]
        valeurs = [d["prix_min_eur_mwh"], d["prix_moyen_eur_mwh"], d["prix_max_eur_mwh"]]
        couleurs = ["#2ca02c", "#1f77b4", "#d62728"]
        fig = go.Figure(data=[go.Bar(x=categories, y=valeurs, marker_color=couleurs, text=[str(v) + " EUR/MWh" for v in valeurs], textposition="auto")])
        fig.update_layout(title="Prix electricite - " + d["periode"], title_x=0.5, yaxis_title="EUR/MWh")
        fig.show()
    
    print(f"Agent #3 - Graphique genere : {type_donnee}")

# AGENT #4 — Interprete et repond
def agent_interpretation(question: str, data: str, type_donnee: str) -> str:
    """Redige une reponse claire en francais avec sources citees."""
    prompt = f"""Tu es un expert en analyse des donnees energetiques francaises.
L utilisateur a pose cette question : {question}
Voici les donnees recuperees : {data}
Redige une reponse claire, structuree et accessible en francais.
- Commence par les chiffres cles
- Explique les tendances
- Cite toujours la source des donnees
- Maximum 5 phrases"""

    response = llm.invoke([HumanMessage(content=prompt)])
    print(f"Agent #4 - Interpretation generee")
    return response.content

# ORCHESTRATEUR intelligent - sans mots-cles fixes
def orchestrateur(question: str) -> str:
    """Coordonne les 5 agents - le LLM decide lui-meme du routage."""
    
    llm_router = ChatOpenAI(
        model="gpt-3.5-turbo",
        temperature=0,
        openai_api_key=os.getenv("OPENAI_API_KEY")
    )
    
    # Le LLM decide lui-meme comment router la question
    prompt_routage = f"""Tu es un assistant specialise dans les donnees energetiques francaises.
Analyse cette question et reponds UNIQUEMENT avec un de ces 3 mots :
- DONNEES : si la question demande des chiffres, statistiques, consommation, mix, prix, production
- CONTEXTE : si la question demande une explication, contexte, historique, strategie, comparaison
- HORSUJET : si la question n a aucun rapport avec l energie

Question : {question}

Reponds avec UN SEUL MOT parmi : DONNEES, CONTEXTE, HORSUJET"""

    routage = llm_router.invoke([HumanMessage(content=prompt_routage)]).content.strip().upper()
    print(f"Routage detecte : {routage}")
    
    if "HORSUJET" in routage:
        return """Je suis specialisee dans les donnees energetiques francaises.
Posez-moi une question sur l energie et je ferai de mon mieux pour vous repondre !"""
    
    elif "CONTEXTE" in routage:
        print("Agent #5 RAG active")
        return agent_rag(question)
    
    else:
        # DONNEES → 4 agents
        params = agent_analyse(question)
        data, type_donnee = agent_donnees(params)
        agent_visualisation(data, type_donnee)
        reponse = agent_interpretation(question, data, type_donnee)
        return reponse

print("Orchestrateur intelligent pret !")
print("Le LLM decide lui-meme du routage — pas de mots-cles fixes !")
```

```python
# Cellule 16 — Interface interactive en boucle
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets

print("Bonjour je suis votre Assistante Energetique Interactif")
print("Posez votre question et appuyez sur Analyser sinon vous pouvez quitter")

# Zone de texte
question_input = widgets.Text(
    placeholder="Ex: Quel est le mix energetique le 15 janvier 2024 ?",
    description="Question :",
    layout=widgets.Layout(width="600px")
)

# Boutons
bouton_analyser = widgets.Button(
    description="Analyser",
    button_style="primary",
    layout=widgets.Layout(width="150px")
)
bouton_quitter = widgets.Button(
    description="Quitter",
    button_style="danger",
    layout=widgets.Layout(width="150px")
)
bouton_nouvelle = widgets.Button(
    description="Nouvelle question",
    button_style="info",
    layout=widgets.Layout(width="180px")
)

# Zone de sortie
output = widgets.Output()

# Compteur de questions
compteur = {"nb": 0}

# Action bouton Analyser
def on_analyser(b):
    with output:
        question = question_input.value.strip()
        
        if question == "":
            print("Veuillez entrer une question !\n")
            return
        
        if question.lower() == "quitter":
            clear_output()
            print(f"Au revoir ! Vous avez pose {compteur['nb']} question(s).")
            return
        
        compteur["nb"] += 1
        #print(f"\nQuestion {compteur['nb']} : {question}")
        #print("-" * 40)
        
        # Utilise l orchestrateur multi-agents
        reponse = orchestrateur(question)
        
        print(f"\nReponse :\n{reponse}")
        print(f"\nTrace : smith.langchain.com")
        print("\nPosez une nouvelle question ou cliquez sur Quitter")

# Action bouton Nouvelle question
def on_nouvelle(b):
    question_input.value = ""

# Action bouton Quitter
def on_quitter(b):
    with output:
        clear_output()
        print(f"Au revoir ! Vous avez pose {compteur['nb']} question(s). A bientot !")

bouton_analyser.on_click(on_analyser)
bouton_nouvelle.on_click(on_nouvelle)
bouton_quitter.on_click(on_quitter)

# Affichage
display(question_input)
display(widgets.HBox([bouton_analyser, bouton_nouvelle, bouton_quitter]))
display(output)
```
